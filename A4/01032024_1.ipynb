### Maze environment

```python
class MiniMaze:
    def __init__(self, size=3):
        self.size = size
        self.terminal_state = (size - 1, size - 1)
        self.state = (0, 0)  # Start at bottom-left corner

    def step(self, action):
        r, c = self.state
        if action == 'up': r = min(r + 1, self.size - 1)
        elif action == 'down': r = max(r - 1, 0)
        elif action == 'left': c = max(c - 1, 0)
        elif action == 'right': c = min(c + 1, self.size - 1)
        self.state = (r, c)
        return self.state, 1 if self.state == self.terminal_state else 0

    def reset(self):
        self.state = (0, 0)
        return self.state
```
